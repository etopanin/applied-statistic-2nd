```python
# Вариант 5 — Y = X7

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error
from statsmodels.stats.outliers_influence import variance_inflation_factor
import statsmodels.api as sm


file1 = "../data/3.xlsx"
file2 = "../data/15.xlsx"
file3 = "../data/19.xlsx"

sheets = {
    "X1": ("3.4.", file1),
    "X2": ("3.2.", file1),
    "X3": ("3.16.", file1),
    "X4": ("3.7.", file1),
    "X5": ("3.13.", file1),
    "X6": ("19.12.", file3),
    "X7": ("3.19.", file1),  
    "X8": ("19.15.1.", file3),
    "X9": ("15.3.", file2),
    "X10": ("15.10.3.", file2),
}

dfs = {}

for key, (sheet, file) in sheets.items():
    df = pd.read_excel(file, sheet_name=sheet, header=6, skiprows=[7])
    df = df.rename(columns={df.columns[0]: "Регион"})
    df = df.set_index("Регион")
    dfs[key] = df



for k in dfs:
    dfs[k].columns = (
        dfs[k].columns.astype(str)          # всё в строки
        .str.strip()                        # убрать пробелы
        .str.replace(r'\.0$', '', regex=True)  # убрать .0
    )
years = ["2021"]

for k in dfs:
    # print(k, dfs[k].columns.tolist())
    dfs[k] = dfs[k][years]
    dfs[k] = dfs[k].apply(pd.to_numeric, errors='coerce')

data = None
for k, df in dfs.items():
    df_melt = df.reset_index().melt(id_vars="Регион", var_name="Год", value_name=k)
    if data is None:
        data = df_melt
    else:
        data = pd.merge(data, df_melt, on=["Регион", "Год"], how="inner")

data.dropna(inplace=True)

print("Объединенные данные:")
display(data)

desc = data.describe().T
print("Основные статистики:")
display(desc)

data[["X1","X2","X3","X4","X5","X6","X7","X8","X9","X10"]].hist(bins=20, figsize=(12,10))
plt.suptitle("Гистограммы признаков")
plt.show()

# Корреляционный анализ 
corr = data.drop(columns=["Регион","Год"]).corr(method='pearson')

plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Матрица корреляции Пирсона")
plt.show()

X = data[[f"X{i}" for i in range(1,11) if i != 7]]
vif_data = pd.DataFrame()
vif_data["Признак"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
print("Фактор инфляции дисперсии (VIF):")
display(vif_data)

Y = data["X7"]  # результирующий признак
X = sm.add_constant(X)
model = sm.OLS(Y, X).fit()
print(model.summary())

residuals = model.resid
fitted = model.fittedvalues

plt.figure(figsize=(6,4))
sns.scatterplot(x=fitted, y=residuals)
plt.axhline(0, color="red", linestyle="--")
plt.title("Остатки vs Предсказанные значения")
plt.xlabel("Предсказанные значения")
plt.ylabel("Остатки")
plt.show()

# Проверка нормальности остатков
sm.qqplot(residuals, line='s')
plt.title("QQ-график остатков")
plt.show()

_, p_shapiro = stats.shapiro(residuals)
print(f"Shapiro-Wilk p-value: {p_shapiro:.4f}")

# Улучшение модели (удаляем незначимые предикторы)
significant = model.pvalues[model.pvalues < 0.05].index.tolist()
if "const" in significant:
    significant.remove("const")

print("Значимые предикторы:", significant)

X_sig = sm.add_constant(data[significant])
model_sig = sm.OLS(Y, X_sig).fit()
print(model_sig.summary())

# Качество модели
y_pred = model_sig.predict(X_sig)

MAE = mean_absolute_error(Y, y_pred)

RMSE = np.sqrt(mean_squared_error(Y, y_pred))
R2 = r2_score(Y, y_pred)
RSE = np.sqrt(np.sum(residuals**2)/(len(Y)-len(X_sig.columns)))
MAPE = np.mean(np.abs((Y - y_pred)/Y))*100

print(f"\nОценки качества модели:")
print(f"MAE = {MAE:.3f}")
print(f"RMSE = {RMSE:.3f}")
print(f"RSE = {RSE:.3f}")
print(f"R² = {R2:.3f}")
print(f"MAPE = {MAPE:.2f}%")

# Прогноз на будущее (при росте факторов на 20%)
median_vals = data[significant].median() * 1.2

future = pd.DataFrame([median_vals], columns=significant)

future = sm.add_constant(future, has_constant='add')

y_future = model_sig.predict(future)
print("\nПрогноз Y (при 120% медианных значений факторов):", y_future.values[0])
```
