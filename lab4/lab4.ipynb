```python
# Вариант 5 — Y = X7

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error
from statsmodels.stats.outliers_influence import variance_inflation_factor
import statsmodels.api as sm


file1 = "../data/3.xlsx"
file2 = "../data/15.xlsx"
file3 = "../data/19.xlsx"

sheets = {
    "X1": ("3.4.", file1),
    "X2": ("3.2.", file1),
    "X3": ("3.16.", file1),
    "X4": ("3.7.", file1),
    "X5": ("3.13.", file1),
    "X6": ("19.12.", file3),
    "X7": ("3.19.", file1),  
    "X8": ("19.15.1.", file3),
    "X9": ("15.3.", file2),
    "X10": ("15.10.3.", file2),
}

dfs = {}

for key, (sheet, file) in sheets.items():
    df = pd.read_excel(file, sheet_name=sheet, header=6, skiprows=[7])
    df = df.rename(columns={df.columns[0]: "Регион"})
    df = df.set_index("Регион")
    dfs[key] = df



for k in dfs:
    dfs[k].columns = (
        dfs[k].columns.astype(str)          # всё в строки
        .str.strip()                        # убрать пробелы
        .str.replace(r'\.0$', '', regex=True)  # убрать .0
    )
years = ["2021"]

for k in dfs:
    # print(k, dfs[k].columns.tolist())
    dfs[k] = dfs[k][years]
    dfs[k] = dfs[k].apply(pd.to_numeric, errors='coerce')

data = None
for k, df in dfs.items():
    df_melt = df.reset_index().melt(id_vars="Регион", var_name="Год", value_name=k)
    if data is None:
        data = df_melt
    else:
        data = pd.merge(data, df_melt, on=["Регион", "Год"], how="inner")

data.dropna(inplace=True)

print("Объединенные данные:")
display(data)

desc = data.describe().T
print("Основные статистики:")
display(desc)

data[["X1","X2","X3","X4","X5","X6","X7","X8","X9","X10"]].hist(bins=20, figsize=(12,10))
plt.suptitle("Гистограммы признаков")
plt.show()

# Корреляционный анализ 
corr = data.drop(columns=["Регион","Год"]).corr(method='pearson')

plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Матрица корреляции Пирсона")
plt.show()

X = data[[f"X{i}" for i in range(1,11) if i != 7]]
vif_data = pd.DataFrame()
vif_data["Признак"] = X.columns
vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
print("Фактор инфляции дисперсии (VIF):")
display(vif_data)

Y = data["X7"]  # результирующий признак
X = sm.add_constant(X)
model = sm.OLS(Y, X).fit()
print(model.summary())

residuals = model.resid
fitted = model.fittedvalues

plt.figure(figsize=(6,4))
sns.scatterplot(x=fitted, y=residuals)
plt.axhline(0, color="red", linestyle="--")
plt.title("Остатки vs Предсказанные значения")
plt.xlabel("Предсказанные значения")
plt.ylabel("Остатки")
plt.show()

# Проверка нормальности остатков
sm.qqplot(residuals, line='s')
plt.title("QQ-график остатков")
plt.show()

_, p_shapiro = stats.shapiro(residuals)
print(f"Shapiro-Wilk p-value: {p_shapiro:.4f}")

# Улучшение модели (удаляем незначимые предикторы)
significant = model.pvalues[model.pvalues < 0.05].index.tolist()
if "const" in significant:
    significant.remove("const")

print("Значимые предикторы:", significant)

X_sig = sm.add_constant(data[significant])
model_sig = sm.OLS(Y, X_sig).fit()
print(model_sig.summary())

# Качество модели
y_pred = model_sig.predict(X_sig)

MAE = mean_absolute_error(Y, y_pred)

RMSE = np.sqrt(mean_squared_error(Y, y_pred))
R2 = r2_score(Y, y_pred)
RSE = np.sqrt(np.sum(residuals**2)/(len(Y)-len(X_sig.columns)))
MAPE = np.mean(np.abs((Y - y_pred)/Y))*100

print(f"\nОценки качества модели:")
print(f"MAE = {MAE:.3f}")
print(f"RMSE = {RMSE:.3f}")
print(f"RSE = {RSE:.3f}")
print(f"R² = {R2:.3f}")
print(f"MAPE = {MAPE:.2f}%")

# Прогноз на будущее (при росте факторов на 20%)
median_vals = data[significant].median() * 1.2

future = pd.DataFrame([median_vals], columns=significant)

future = sm.add_constant(future, has_constant='add')

y_future = model_sig.predict(future)
print("\nПрогноз Y (при 120% медианных значений факторов):", y_future.values[0])
```

```python
# Вариант 5 — Y = X7

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error
from statsmodels.stats.outliers_influence import variance_inflation_factor
import statsmodels.api as sm

print("="*60)
print("ПОСТРОЕНИЕ И ОПТИМИЗАЦИЯ РЕГРЕССИОННОЙ МОДЕЛИ")
print("="*60)

# [Ваш код загрузки данных остается без изменений]
file1 = "../data/3.xlsx"
file2 = "../data/15.xlsx"
file3 = "../data/19.xlsx"

sheets = {
    "X1": ("3.4.", file1),
    "X2": ("3.2.", file1),
    "X3": ("3.16.", file1),
    "X4": ("3.7.", file1),
    "X5": ("3.13.", file1),
    "X6": ("19.12.", file3),
    "X7": ("3.19.", file1),  
    "X8": ("19.15.1.", file3),
    "X9": ("15.3.", file2),
    "X10": ("15.10.3.", file2),
}

dfs = {}

for key, (sheet, file) in sheets.items():
    df = pd.read_excel(file, sheet_name=sheet, header=6, skiprows=[7])
    df = df.rename(columns={df.columns[0]: "Регион"})
    df = df.set_index("Регион")
    dfs[key] = df

for k in dfs:
    dfs[k].columns = (
        dfs[k].columns.astype(str)
        .str.strip()
        .str.replace(r'\.0$', '', regex=True)
    )
years = ["2021"]

for k in dfs:
    dfs[k] = dfs[k][years]
    dfs[k] = dfs[k].apply(pd.to_numeric, errors='coerce')

data = None
for k, df in dfs.items():
    df_melt = df.reset_index().melt(id_vars="Регион", var_name="Год", value_name=k)
    if data is None:
        data = df_melt
    else:
        data = pd.merge(data, df_melt, on=["Регион", "Год"], how="inner")

data.dropna(inplace=True)

print("Размерность данных после обработки:", data.shape)
print("\nПервые 5 строк данных:")
display(data.head())

# =============================================================================
# 1. ПОСТРОЕНИЕ ИСХОДНОЙ МОДЕЛИ СО ВСЕМИ ПРЕДИКТОРАМИ
# =============================================================================

print("\n" + "="*60)
print("1. ИСХОДНАЯ МОДЕЛЬ СО ВСЕМИ ПРЕДИКТОРАМИ")
print("="*60)

Y = data["X7"]
X_full = data[[f"X{i}" for i in range(1,11) if i != 7]]
X_full = sm.add_constant(X_full)

model_full = sm.OLS(Y, X_full).fit()
print("МОДЕЛЬ СО ВСЕМИ ПРЕДИКТОРАМИ:")
print(model_full.summary())

# Анализ значимости коэффициентов
print("\nСТАТИСТИЧЕСКАЯ ЗНАЧИМОСТЬ КОЭФФИЦИЕНТОВ:")
pvalues = model_full.pvalues
for var, pval in pvalues.items():
    significance = "✓ ЗНАЧИМ" if pval < 0.05 else "✗ НЕЗНАЧИМ"
    print(f"{var}: p-value = {pval:.4f} {significance}")

# =============================================================================
# 2. ПОШАГОВОЕ ИСКЛЮЧЕНИЕ НЕЗНАЧИМЫХ ПРЕДИКТОРОВ
# =============================================================================

print("\n" + "="*60)
print("2. ПОШАГОВОЕ ИСКЛЮЧЕНИЕ НЕЗНАЧИМЫХ ПРЕДИКТОРОВ")
print("="*60)

def stepwise_backward_elimination(X, Y, significance_level=0.05):
    """Пошаговое исключение незначимых предикторов"""
    predictors = X.columns.tolist()
    if 'const' in predictors:
        predictors.remove('const')
    
    current_predictors = predictors.copy()
    history = []
    
    print("Начало пошагового исключения...")
    
    while True:
        # Строим модель с текущими предикторами
        X_current = sm.add_constant(X[current_predictors])
        model = sm.OLS(Y, X_current).fit()
        
        # Находим самый незначимый предиктор (кроме константы)
        pvalues = model.pvalues.drop('const', errors='ignore')
        if len(pvalues) == 0:
            break
            
        max_pval = pvalues.max()
        worst_predictor = pvalues.idxmax()
        
        # Если все предикторы значимы - останавливаемся
        if max_pval < significance_level:
            print("Все оставшиеся предикторы статистически значимы")
            break
        
        # Удаляем самый незначимый предиктор
        current_predictors.remove(worst_predictor)
        
        # Сохраняем историю
        history.append({
            'removed': worst_predictor,
            'p_value': max_pval,
            'r_squared': model.rsquared,
            'adj_r_squared': model.rsquared_adj,
            'predictors': current_predictors.copy()
        })
        
        print(f"Удален: {worst_predictor} (p-value = {max_pval:.4f})")
        print(f"Оставшиеся предикторы: {current_predictors}")
        print(f"R² = {model.rsquared:.4f}, Adj R² = {model.rsquared_adj:.4f}")
        print("-" * 50)
        
        if not current_predictors:
            break
    
    return current_predictors, history

# Запускаем пошаговое исключение
predictors_to_keep, elimination_history = stepwise_backward_elimination(
    data[[f"X{i}" for i in range(1,11) if i != 7]], Y
)

print(f"\nФИНАЛЬНЫЙ НАБОР ЗНАЧИМЫХ ПРЕДИКТОРОВ: {predictors_to_keep}")

# =============================================================================
# 3. ПОСТРОЕНИЕ УЛУЧШЕННОЙ МОДЕЛИ
# =============================================================================

print("\n" + "="*60)
print("3. УЛУЧШЕННАЯ МОДЕЛЬ СО ЗНАЧИМЫМИ ПРЕДИКТОРАМИ")
print("="*60)

if predictors_to_keep:
    X_optimized = sm.add_constant(data[predictors_to_keep])
    model_optimized = sm.OLS(Y, X_optimized).fit()
    
    print("УЛУЧШЕННАЯ МОДЕЛЬ:")
    print(model_optimized.summary())
else:
    print("НЕТ ЗНАЧИМЫХ ПРЕДИКТОРОВ!")
    # Создаем минимальную модель с константой
    X_optimized = sm.add_constant(pd.DataFrame(index=Y.index))
    model_optimized = sm.OLS(Y, X_optimized).fit()

# =============================================================================
# 4. СРАВНЕНИЕ КАЧЕСТВА МОДЕЛЕЙ
# =============================================================================

print("\n" + "="*60)
print("4. СРАВНЕНИЕ КАЧЕСТВА МОДЕЛЕЙ")
print("="*60)

# Прогнозы для обеих моделей
y_pred_full = model_full.predict(X_full)
y_pred_optimized = model_optimized.predict(X_optimized)

# Метрики для полной модели
mae_full = mean_absolute_error(Y, y_pred_full)
rmse_full = np.sqrt(mean_squared_error(Y, y_pred_full))
r2_full = r2_score(Y, y_pred_full)
rse_full = np.sqrt(np.sum((Y - y_pred_full)**2) / (len(Y) - X_full.shape[1]))
mape_full = np.mean(np.abs((Y - y_pred_full) / Y)) * 100

# Метрики для оптимизированной модели
mae_opt = mean_absolute_error(Y, y_pred_optimized)
rmse_opt = np.sqrt(mean_squared_error(Y, y_pred_optimized))
r2_opt = r2_score(Y, y_pred_optimized)
rse_opt = np.sqrt(np.sum((Y - y_pred_optimized)**2) / (len(Y) - X_optimized.shape[1]))
mape_opt = np.mean(np.abs((Y - y_pred_optimized) / Y)) * 100

comparison = pd.DataFrame({
    'Полная модель': [
        len(model_full.params) - 1,
        model_full.rsquared,
        model_full.rsquared_adj,
        mae_full,
        rmse_full,
        rse_full,
        mape_full,
        model_full.f_pvalue
    ],
    'Оптимизированная модель': [
        len(model_optimized.params) - 1,
        model_optimized.rsquared,
        model_optimized.rsquared_adj,
        mae_opt,
        rmse_opt,
        rse_opt,
        mape_opt,
        model_optimized.f_pvalue
    ]
}, index=['Число предикторов', 'R²', 'Adj R²', 'MAE', 'RMSE', 'RSE', 'MAPE (%)', 'F-test p-value'])

print("СРАВНЕНИЕ МОДЕЛЕЙ:")
display(comparison.round(4))

# =============================================================================
# 5. ПРОВЕРКА ПРЕДПОСЫЛОК ДЛЯ УЛУЧШЕННОЙ МОДЕЛИ
# =============================================================================

print("\n" + "="*60)
print("5. ПРОВЕРКА ПРЕДПОСЫЛОК МНК ДЛЯ УЛУЧШЕННОЙ МОДЕЛИ")
print("="*60)

residuals_opt = model_optimized.resid
fitted_opt = model_optimized.fittedvalues

# 5.1 График остатков vs предсказанные значения
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
sns.scatterplot(x=fitted_opt, y=residuals_opt)
plt.axhline(0, color="red", linestyle="--")
plt.title("Остатки vs Предсказанные значения")
plt.xlabel("Предсказанные значения")
plt.ylabel("Остатки")

# 5.2 Q-Q график
plt.subplot(1, 3, 2)
sm.qqplot(residuals_opt, line='s', ax=plt.gca())
plt.title("Q-Q график остатков")

# 5.3 Гистограмма остатков
plt.subplot(1, 3, 3)
sns.histplot(residuals_opt, kde=True)
plt.title("Распределение остатков")
plt.xlabel("Остатки")

plt.tight_layout()
plt.show()

# 5.4 Статистические тесты
print("\nСТАТИСТИЧЕСКИЕ ТЕСТЫ ДЛЯ УЛУЧШЕННОЙ МОДЕЛИ:")

# Тест на нормальность остатков (Шапиро-Уилк)
shapiro_stat, shapiro_p = stats.shapiro(residuals_opt)
print(f"Тест Шапиро-Уилка на нормальность: W = {shapiro_stat:.4f}, p = {shapiro_p:.4f}")

# Тест на гомоскедастичность (Бройша-Пагана)
bp_test = sm.stats.het_breuschpagan(residuals_opt, model_optimized.model.exog)
print(f"Тест Бройша-Пагана на гомоскедастичность: LM = {bp_test[0]:.4f}, p = {bp_test[1]:.4f}")

# Тест на автокорреляцию (Дарбина-Уотсона)
dw_stat = sm.stats.stattools.durbin_watson(residuals_opt)
print(f"Тест Дарбина-Уотсона на автокорреляцию: DW = {dw_stat:.4f}")

# =============================================================================
# 6. ВЫВОДЫ И РЕКОМЕНДАЦИИ
# =============================================================================

print("\n" + "="*60)
print("6. ВЫВОДЫ И РЕКОМЕНДАЦИИ")
print("="*60)

print("\nСТАТИСТИЧЕСКАЯ ЗНАЧИМОСТЬ МОДЕЛИ:")
if model_optimized.f_pvalue < 0.05:
    print("✓ Модель в целом статистически значима")
else:
    print("✗ Модель в целом незначима")

print("\nЗНАЧИМОСТЬ КОЭФФИЦИЕНТОВ:")
significant_coefs = model_optimized.pvalues[model_optimized.pvalues < 0.05]
if len(significant_coefs) > 0:
    print("Значимые коэффициенты:")
    for coef, pval in significant_coefs.items():
        print(f"  {coef}: p-value = {pval:.4f}")
else:
    print("Нет статистически значимых коэффициентов")

print(f"\nКАЧЕСТВО МОДЕЛИ:")
print(f"R² = {model_optimized.rsquared:.4f} ({model_optimized.rsquared*100:.1f}% объясненной дисперсии)")
print(f"Скорректированный R² = {model_optimized.rsquared_adj:.4f}")

print(f"\nТОЧНОСТЬ ПРОГНОЗА:")
print(f"Средняя абсолютная ошибка (MAE): {mae_opt:.2f}")
print(f"Средняя абсолютная процентная ошибка (MAPE): {mape_opt:.2f}%")

print(f"\nПРЕДПОСЫЛКИ МНК:")
print(f"Нормальность остатков: {'СОБЛЮДЕНА' if shapiro_p > 0.05 else 'НАРУШЕНА'}")
print(f"Гомоскедастичность: {'СОБЛЮДЕНА' if bp_test[1] > 0.05 else 'НАРУШЕНА'}")
print(f"Автокорреляция: {'НЕТ' if 1.5 < dw_stat < 2.5 else 'ВОЗМОЖНО ЕСТЬ'}")

# =============================================================================
# 7. ПРОГНОЗ НА ОСНОВЕ УЛУЧШЕННОЙ МОДЕЛИ
# =============================================================================

if predictors_to_keep:
    print("\n" + "="*60)
    print("7. ПРОГНОЗИРОВАНИЕ")
    print("="*60)
    
    # Прогноз при 120% медианных значений
    median_vals = data[predictors_to_keep].median() * 1.2
    future = pd.DataFrame([median_vals], columns=predictors_to_keep)
    future = sm.add_constant(future, has_constant='add')
    
    y_future = model_optimized.predict(future)
    print(f"Прогноз Y (при 120% медианных значений факторов): {y_future.values[0]:.2f}")
    
    # Доверительный интервал прогноза
    prediction = model_optimized.get_prediction(future)
    ci = prediction.conf_int(alpha=0.05)
    print(f"95% доверительный интервал: [{ci[0,0]:.2f}, {ci[0,1]:.2f}]")

print("\n" + "="*60)
print("АНАЛИЗ ЗАВЕРШЕН")
print("="*60)
```
